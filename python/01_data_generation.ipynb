```python
np.random.seed(42)

campaign_creator_rows = []

for campaign_id in campaigns["campaign_id"]:
    campaign_category = campaigns.loc[
        campaigns["campaign_id"] == campaign_id,
        "campaign_category"
    ].iloc[0]

    matching_creators = creators[
        creators["category"] == campaign_category
    ]["creator_id"].tolist()

    selected_creators = np.random.choice(
        matching_creators,
        size=min(3, len(matching_creators)),
        replace=False
    )

    for creator_id in selected_creators:
        campaign_budget = campaigns.loc[
            campaigns["campaign_id"] == campaign_id,
            "budget"
        ].iloc[0]

        campaign_creator_rows.append({
            "campaign_id": campaign_id,
            "creator_id": int(creator_id),
            "agreed_cost": round(
                campaign_budget * np.random.uniform(0.15, 0.35), 2
            )
        })

campaign_creators = pd.DataFrame(campaign_creator_rows)

campaign_creators.head()
```

```python
np.random.seed(42)

performance_rows = []

for _, row in campaign_creators.iterrows():

    views = np.random.randint(50000, 1000000)

    likes = int(views * np.random.uniform(0.03, 0.10))
    comments = int(views * np.random.uniform(0.001, 0.015))
    clicks = int(views * np.random.uniform(0.01, 0.08))
    conversions = int(clicks * np.random.uniform(0.03, 0.15))

    revenue = round(
        conversions * np.random.uniform(500, 3000),
        2
    )

    performance_rows.append({
        "campaign_id": row["campaign_id"],
        "creator_id": row["creator_id"],
        "views": views,
        "likes": likes,
        "comments": comments,
        "clicks": clicks,
        "conversions": conversions,
        "revenue": revenue
    })

campaign_performance = pd.DataFrame(performance_rows)

campaign_performance.head()
```

```python
creators = pd.read_csv("../data/generated/creators.csv")
channels = pd.read_csv("../data/generated/channels.csv")
videos = pd.read_csv("../data/generated/videos.csv")
video_metrics = pd.read_csv("../data/generated/video_metrics.csv")

print("Scoring datasets loaded successfully! ✅")
```

```python
creator_base = creators.merge(
    channels,
    on="creator_id",
    how="inner"
)

print("Creator base created successfully! ✅")
print(creator_base.head())
```

```python
creator_base["avg_views_per_video"] = (
    creator_base["total_views"] / creator_base["total_videos"]
).round(0)

print(creator_base[
    ["creator_name", "subscribers", "avg_views_per_video"]
].head(10))
```

```python
video_data = videos.merge(
    video_metrics,
    on="video_id",
    how="inner"
)

creator_engagement = (
    video_data
    .merge(
        channels[["channel_id", "creator_id"]],
        on="channel_id",
        how="inner"
    )
    .groupby("creator_id")
    .agg(
        total_video_views=("views", "sum"),
        total_likes=("likes", "sum"),
        total_comments=("comments", "sum")
    )
    .reset_index()
)

creator_engagement["engagement_rate"] = (
    (creator_engagement["total_likes"] +
     creator_engagement["total_comments"])
    / creator_engagement["total_video_views"] * 100
).round(2)

print(creator_engagement.head(10))
```

```python
creator_base = creator_base.merge(
    creator_engagement[
        ["creator_id", "total_video_views", "total_likes",
         "total_comments", "engagement_rate"]
    ],
    on="creator_id",
    how="left"
)

print(
    creator_base[
        [
            "creator_name",
            "subscribers",
            "avg_views_per_video",
            "engagement_rate"
        ]
    ].head(10)
)
```

```python
creator_base["views_score"] = (
    creator_base["avg_views_per_video"]
    / creator_base["avg_views_per_video"].max()
    * 100
).round(2)

print(
    creator_base[
        ["creator_name", "avg_views_per_video", "views_score"]
    ].sort_values("views_score", ascending=False).head(10)
)
```

```python
creator_base["engagement_score"] = (
    creator_base["engagement_rate"]
    / creator_base["engagement_rate"].max()
    * 100
).round(2)

print(
    creator_base[
        ["creator_name", "engagement_rate", "engagement_score"]
    ].sort_values("engagement_score", ascending=False).head(10)
)
```

```python
creator_base["subscriber_score"] = (
    creator_base["subscribers"]
    / creator_base["subscribers"].max()
    * 100
).round(2)

print(
    creator_base[
        ["creator_name", "subscribers", "subscriber_score"]
    ].sort_values("subscriber_score", ascending=False).head(10)
)
```

```python
creator_base["creator_performance_score"] = (
    creator_base["engagement_score"] * 0.40
    + creator_base["views_score"] * 0.35
    + creator_base["subscriber_score"] * 0.25
).round(2)

print(
    creator_base[
        [
            "creator_name",
            "category",
            "views_score",
            "engagement_score",
            "subscriber_score",
            "creator_performance_score"
        ]
    ]
    .sort_values("creator_performance_score", ascending=False)
    .head(10)
)
```

```python
creator_score = creator_base[
    [
        "creator_id",
        "creator_name",
        "category",
        "location",
        "language",
        "subscribers",
        "avg_views_per_video",
        "engagement_rate",
        "creator_performance_score"
    ]
].sort_values(
    "creator_performance_score",
    ascending=False
)

creator_score.to_csv(
    "../data/generated/creator_performance_score.csv",
    index=False
)

print("Creator performance score saved successfully! ✅")
```

```python
campaign_creators = pd.read_csv(
    "../data/generated/campaign_creators.csv"
)

campaign_performance = pd.read_csv(
    "../data/generated/campaign_performance.csv"
)

print("Campaign datasets loaded successfully! ✅")
```

```python
creator_campaign = campaign_creators.merge(
    campaign_performance,
    on=["campaign_id", "creator_id"],
    how="inner"
)

print(creator_campaign.head())
```

```python
campaign_summary = (
    creator_campaign
    .groupby("creator_id")
    .agg(
        total_campaigns=("campaign_id", "nunique"),
        campaign_views=("views", "sum"),
        campaign_clicks=("clicks", "sum"),
        campaign_conversions=("conversions", "sum"),
        campaign_revenue=("revenue", "sum"),
        campaign_cost=("agreed_cost", "sum")
    )
    .reset_index()
)

print(campaign_summary.head(10))
```

```python
campaign_summary["campaign_conversion_rate"] = (
    campaign_summary["campaign_conversions"]
    / campaign_summary["campaign_clicks"]
    * 100
).round(2)

print(
    campaign_summary[
        [
            "creator_id",
            "campaign_clicks",
            "campaign_conversions",
            "campaign_conversion_rate"
        ]
    ].sort_values(
        "campaign_conversion_rate",
        ascending=False
    ).head(10)
)
```

```python
campaign_summary["campaign_score"] = (
    campaign_summary["campaign_conversion_rate"]
    / campaign_summary["campaign_conversion_rate"].max()
    * 100
).round(2)

print(
    campaign_summary[
        [
            "creator_id",
            "campaign_conversion_rate",
            "campaign_score"
        ]
    ].sort_values(
        "campaign_score",
        ascending=False
    ).head(10)
)
```

```python
creator_base = creator_base.merge(
    campaign_summary[
        [
            "creator_id",
            "total_campaigns",
            "campaign_views",
            "campaign_clicks",
            "campaign_conversions",
            "campaign_revenue",
            "campaign_cost",
            "campaign_conversion_rate",
            "campaign_score"
        ]
    ],
    on="creator_id",
    how="left"
)

print(
    creator_base[
        [
            "creator_name",
            "creator_performance_score",
            "campaign_conversion_rate",
            "campaign_score"
        ]
    ].head(10)
)
```

```python
creator_base["creatorbridge_score"] = (
    creator_base["creator_performance_score"] * 0.70
    + creator_base["campaign_score"] * 0.30
).round(2)

print(
    creator_base[
        [
            "creator_name",
            "category",
            "creator_performance_score",
            "campaign_score",
            "creatorbridge_score"
        ]
    ]
    .sort_values(
        "creatorbridge_score",
        ascending=False
    )
    .head(10)
)
```

```python
final_creator_scores = creator_base[
    [
        "creator_id",
        "creator_name",
        "category",
        "location",
        "language",
        "subscribers",
        "avg_views_per_video",
        "engagement_rate",
        "total_campaigns",
        "campaign_views",
        "campaign_clicks",
        "campaign_conversions",
        "campaign_conversion_rate",
        "creator_performance_score",
        "campaign_score",
        "creatorbridge_score"
    ]
].sort_values(
    "creatorbridge_score",
    ascending=False
)

final_creator_scores.to_csv(
    "../data/generated/final_creator_scores.csv",
    index=False
)

print("Final CreatorBridge scores saved successfully! ✅")
```
